# SAE text detector (DetectorScope) — Colab sketch

CPU-friendly walkthrough of the tech-demos `apps/sae-text-detector` kit.
Uses the official [`detectorscope`](https://github.com/google-deepmind/detectorscope) package.
**GPU is optional and not required.** Do not use paid runtimes without owner approval.

Paper: [arXiv:2610.30006](https://arxiv.org/abs/2610.30006).

```python
# !pip install -q detectorscope gradio
from detectorscope import AuditableDetector

det = AuditableDetector.from_pretrained("liamdugan/detector-scope")
text = (
    "In this paper, we propose a novel framework that leverages synergistic "
    "paradigms to optimize holistic outcomes across diverse domains."
)
out = det(text, tau=0.85, return_highlights=True)
print(out)
print("top features:")
for f in out.features[:8]:
    print(f"  {f.importance:+.3f}  {f.description}")
```
